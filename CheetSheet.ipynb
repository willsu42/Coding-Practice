**Two pointers: one input, opposite ends**

```python
def fn(arr):
    left = ans = 0
    right = len(arr) - 1

    while left < right:
        # do some logic here with left and right
        if CONDITION:
            left += 1
        else:
            right -= 1
    
    return ans
```

It’s one of the most efficient ways to search for a pair of elements in a sorted array or solve problems involving palindromes and "trapping" logic. Instead of using a nested loop (which would be O($n^2$)), this approach allows you to process the data in a single pass (O(n)).

Real-World Example: Merging Sorted Lists

```python
def merge_sorted(arr1, arr2):
    i = j = 0
    merged = []

    # Phase 1: Comparison
    while i < len(arr1) and j < len(arr2):
        if arr1[i] < arr2[j]:
            merged.append(arr1[i])
            i += 1
        else:
            merged.append(arr2[j])
            j += 1
    
    # Phase 2: Cleanup (only one of these will actually run)
    merged.extend(arr1[i:])
    merged.extend(arr2[j:])
    
    return merged
```
